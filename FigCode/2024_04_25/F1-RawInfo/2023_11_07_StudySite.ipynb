```python
import geemap 
import ee
ee.Initialize()
```

```python
brazil = ee.FeatureCollection("FAO/GAUL/2015/level0").filter(ee.Filter.eq('ADM0_NAME', 'Brazil'))
```

```python

Ls = ee.ImageCollection("projects/sat-io/open-datasets/GRWL/water_mask_v01_01").filterBounds(brazil)
 #ls data
```

```python
Map = geemap.Map()
Map.addLayer(Ls, {}, 'Water')
Map.centerObject(Ls)
Map
```
